```python
#  remove all clips over 5s and under 0.6s
# remove ones with max volume <threshold?

import shutil

VOLUME_THRESHOLD = 0.001

short_clips = 0
long_clips = 0
inaudible_clips = 0
unknowns = 0

INAUDIBLE_DIR = "inaudible_clips"
shutil.rmtree(INAUDIBLE_DIR, ignore_errors=True)
os.makedirs(INAUDIBLE_DIR, exist_ok=True)

print(f"df len: {len(df)}")
total = 0

to_drop = []  # indices to drop in df

for i, record in df.iterrows():

    file_path = record['file_path']

    try:
        info = sf.info(file_path)
        duration = info.duration

        if duration < 0.6:
            to_drop.append(i)
            short_clips += 1

        elif duration > 5.0:
            to_drop.append(i)
            long_clips += 1

        else:
            data, sr = sf.read(file_path, always_2d=True)
            mono_data = data.mean(axis=1)
            peak = float(np.max(np.abs(mono_data)))

            if peak < VOLUME_THRESHOLD:
                to_drop.append(i)
                inaudible_clips += 1

                # copy to folder to listen
                dest = os.path.join(INAUDIBLE_DIR, os.path.basename(record['file_path']))
                shutil.copy2(record['file_path'], dest)

    except:
        df = df.drop(index=i)
        unknowns += 1
    
    total += 1


df = df.drop(index=to_drop)
df = df.reset_index(drop=True)

print(f"Total clips processed: {total}\n")

print(f"Long clips removed >5s: {long_clips}")
print(f"Short clips removed <0.6s: {short_clips}")
print(f"Inaudible clips removed: {inaudible_clips}")
print(f"Unkown clips removed ???: {unknowns}\n")

print(f"NEW df len: {len(df)}")

df.head()
```

```python
# preprocess - pad to 5s, resample to fixed sample rate

def preprocess_audio(file_path, target_sr=22050, target_duration=5.0):

    data, sr = sf.read(file_path, dtype='float32')

    # convert to mono
    if data.ndim > 1:
        data = data.mean(axis=1) # avg across channels

    # convert to tensor
    waveform = torch.from_numpy(data).float().unsqueeze(0)  # (1, samples)

    # resample
    if sr != target_sr:
        resampler = T.Resample(sr, target_sr)
        waveform = resampler(waveform)
        #data = librosa.resample(data, orig_sr=sr, target_sr=target_sr)
    
    # padding
    target_samples = int(target_sr * target_duration)
    curr_samples = waveform.shape[1]

    if curr_samples < target_samples:
        padding = target_samples - curr_samples
        left = padding // 2
        right = padding - left
        waveform = F.pad(waveform, (left, right)) # if odd pad 1 extra on right

    elif curr_samples > target_samples:
        print(f"Error - encountered longer file???")

    # Convert to PyTorch tensor with shape (1, samples)
    # Why: Need tensor for PyTorch training, shape (1, samples) = (channels, time)
    #waveform = torch.from_numpy(data).float().unsqueeze(0)

    return waveform
```

```python

def visualise_spectrogram(path=None, waveform=None, sr=22050):
    
    if path:
        data, sr = sf.read(path)
        waveform = torch.from_numpy(data).float()
        
        if waveform.ndim > 1: 
            waveform = waveform.mean(dim=1)  # mono
            
        waveform = waveform.unsqueeze(0)  # (1, samples)
    
    # else waveform is already in right shape

    mel_db = T.AmplitudeToDB()(T.MelSpectrogram(sr, n_fft=1024, hop_length=256, n_mels=64)(waveform))
    print("Tensor shape (C, n_mels, T):", mel_db.shape)

    plt.imshow(mel_db[0].numpy(), origin='lower', aspect='auto', cmap='inferno')
    plt.colorbar(label='dB')
    plt.show()
```

```python
# WAVEFORM AUGMENTATIONS


def add_background_noise(waveform, noise_dir='../background_noise', snr_db_range=(5, 20)):
    """
    Add REAL background noise from audio files.
    Best approach: collect actual environmental noise recordings.
    
    Create a folder with recordings of:
    - Traffic noise
    - Wind
    - Crowd noise
    - Indoor ambience
    """
    import random
    import os
    
    if not os.path.exists(noise_dir):
        print("Not Found!")
        return waveform  # Fall back if no noise files
    
    noise_files = [f for f in os.listdir(noise_dir) if f.endswith(('.wav', '.mp3'))]
    if not noise_files:
        return waveform
    
    # Load random noise file
    noise_file = random.choice(noise_files)
    noise_path = os.path.join(noise_dir, noise_file)
    noise_data, noise_sr = sf.read(noise_path, dtype='float32')
    
    if noise_data.ndim > 1:
        noise_data = noise_data.mean(axis=1)
    
    noise_tensor = torch.from_numpy(noise_data).float().unsqueeze(0)
    
    # Resample if needed
    if noise_sr != 22050:
        resampler = T.Resample(noise_sr, 22050)
        noise_tensor = resampler(noise_tensor)
    
    # Random segment if noise is longer
    if noise_tensor.shape[1] > waveform.shape[1]:
        start = torch.randint(0, noise_tensor.shape[1] - waveform.shape[1], (1,)).item()
        noise_tensor = noise_tensor[:, start:start + waveform.shape[1]]
    elif noise_tensor.shape[1] < waveform.shape[1]:
        # Repeat noise if too short
        repeats = (waveform.shape[1] // noise_tensor.shape[1]) + 1
        noise_tensor = noise_tensor.repeat(1, repeats)[:, :waveform.shape[1]]
    
    # Apply SNR
    snr_db = torch.FloatTensor(1).uniform_(*snr_db_range).item()
    signal_power = waveform.pow(2).mean()
    noise_power = noise_tensor.pow(2).mean()
    snr_linear = 10 ** (snr_db / 10)
    
    scale = torch.sqrt(signal_power / (snr_linear * noise_power))
    noise_tensor = noise_tensor * scale
    
    return waveform + noise_tensor


def time_shift(waveform, target_sr=22050):
    shift = int(torch.randint(-target_sr, target_sr, (1,)))
    waveform = torch.roll(waveform, shift, dims=1)
    return waveform


'''
def add_noise(waveform):
    noise = torch.randn_like(waveform) * 0.001
    waveform += noise
    return waveform
'''

def gain(waveform):
    gain = torch.FloatTensor(1).uniform_(0.5, 2) # 0.7, 1.3
    waveform = waveform * gain
    return waveform


def time_stretch(waveform):
    rate = torch.FloatTensor(1).uniform_(0.6, 1.5).item()       #0.9 , 1.1
    waveform_np = waveform.squeeze(0).numpy()
    stretched = librosa.effects.time_stretch(waveform_np, rate=rate)
    target_len = waveform.shape[1]

    if len(stretched) < target_len:
        stretched = np.pad(stretched, (0, target_len - len(stretched)))
    else:
        stretched = stretched[:target_len]
    waveform = torch.from_numpy(stretched).unsqueeze(0)
    return waveform


def pitch_shift(waveform, target_sr=22050):
    n_steps = torch.FloatTensor(1).uniform_(-2, 2).item()  # ±2 semitones
    waveform_np = waveform.squeeze(0).numpy()
    shifted = librosa.effects.pitch_shift(waveform_np, sr=target_sr, n_steps=n_steps)
    waveform = torch.from_numpy(shifted).unsqueeze(0)
    return waveform



def apply_augmentations(waveform):

    waveform = waveform.clone()

    mask = np.random.uniform(0,1,size=5) # random probabilities - decide activation

    # time shift
    if mask[0] > 0.5:
        waveform = time_shift(waveform)
    
    '''
    # noise
    if mask[1] > 0.5:
        #waveform = add_noise(waveform)
        waveform = add_background_noise(waveform)
    '''
        
    # gain
    if mask[2] > 0.5:
        waveform = gain(waveform)
    
    # time stretch
    if mask[3] > 0.5:
        waveform = time_stretch(waveform)
    
    # pitch shift
    if mask[4] > 0.5:
        waveform = pitch_shift(waveform)
    
    return waveform
```

```python
# SPECTROGRAM AUGMENTATIONS

def frequency_mask(spec, num_masks, mask_params):

    spec = spec.clone()
    num_mel_chanels = spec.shape[1]     # (1, n_mels, time)

    for _ in range(num_masks):
        f = torch.randint(0, mask_params, (1,)).item()
        f0 = torch.randint(0, num_mel_chanels - f, (1,)).item()
        spec[:, f0:f0+f, :] = 0
    
    return spec


def time_mask(spec, num_masks, mask_params):

    spec = spec.clone()
    num_time_steps = spec.shape[2]      # (1, n_mels, time)

    for _ in range(num_masks):
        t = torch.randint(0, mask_params, (1,)).item()
        t0 = torch.randint(0, num_time_steps - t, (1,)).item()
        spec[:, :, t0:t0+t] = 0

    return spec


```

```python
class AudioDataset(Dataset):

    def __init__(self, dataframe, target_sr=22050, target_duration=5.0, n_mels=64,
                 n_fft=1024, hop_length=256, augment=False, n_augmentations=None):
        
        self.df = dataframe.reset_index(drop=True) # why reset?
        self.target_sr = target_sr
        self.target_duration = target_duration
        self.augment = augment
        self.n_augmentations = n_augmentations

        self.mel_transform = T.MelSpectrogram(
            sample_rate=target_sr,
            n_fft=n_fft,
            hop_length=hop_length,
            n_mels=n_mels
        )

        self.db_transform = T.AmplitudeToDB()

        # CREATE LABEL ENCODING
        self.classes = sorted(self.df['class'].unique())
        self.class_to_idx = {cls: idx for idx, cls in enumerate(self.classes)} # dict

        if self.augment:
            total_samples = len(self.df) * (1 + self.n_augmentations)
            print(f"Dataset created with {len(self.df)} original samples")
            print(f"With augmentation: {total_samples} total samples (1 original + {self.n_augmentations} augmented per sample)")
        else:
            print(f"Dataset created with {len(self.df)} samples (no augmentation)")
    
        print(f"Classes: {self.classes}")
        print(f"Class mapping: {self.class_to_idx}\n")
        
    
    def __len__(self):

        if self.augment:
            return len(self.df) * (1 + self.n_augmentations)
        else:
            return len(self.df)
    

    def __getitem__(self, idx):

        """
        Load and process a single sample.
        
        Returns:
            spectrogram: tensor of shape (1, n_mels, time_steps)
            label: integer class label
        """

        if self.augment:
            real_idx = idx // (1 + self.n_augmentations) # original sample which has file path
            aug_variant = idx % (1 + self.n_augmentations) # which version (0=original)
        else:
            real_idx = idx
            aug_variant = 0

        row = self.df.iloc[real_idx]
        file_path = row['file_path']
        cls = row['class']
        label = self.class_to_idx[cls]

        waveform = preprocess_audio(file_path, self.target_sr, self.target_duration) 

        # WAVEFORM AUGMENTATIONS
        if self.augment and aug_variant > 0:
            waveform = apply_augmentations(waveform, self.target_sr)

            if cls != "other" and (np.random.uniform(0,1) > 0.5):   # dont apply bg noise to other sounds (already bg)
                waveform = add_background_noise(waveform)       


        # convert to ???
        mel_spec = self.mel_transform(waveform)
        mel_db = self.db_transform(mel_spec)


        # SPECTROGRAM AUGMENTATIONS
        if self.augment and aug_variant > 0:

            if np.random.uniform(0,1) > 0.5:
                mel_db = frequency_mask(mel_db, num_masks=2, mask_params=15)

            if np.random.uniform(0,1) > 0.5:
                mel_db = time_mask(mel_db, num_masks=2, mask_params=50)


        # NORMALIZATION
        #mu = mel_db.mean()
        #sd = mel_db.std()
        #mel_db = (mel_db - mu) / (sd + 1e-8)    # epsilon added in case of division by 0

        return mel_db, label


```

```python
train_dataset = AudioDataset(train_df, augment=True, n_augmentations=1)     # augments
val_dataset = AudioDataset(val_df, augment=False)
test_dataset = AudioDataset(test_df, augment=False)

train_loader = DataLoader(train_dataset, batch_size=128, shuffle=True, drop_last=True)   # new - drop last batch as its likely to be smaller, can make grad noisy
val_loader = DataLoader(val_dataset, batch_size=128, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=128, shuffle=False)
```

```python
class AudioCNN(nn.Module):

    def __init__(self, num_classes=3, n_mels=64):
        super(AudioCNN, self).__init__()

        # input (batch, 1, n_mels, time)

        # Conv block 1
        self.conv1 = nn.Conv2d(1, 64, kernel_size=3, padding=1)
        self.bn1 = nn.BatchNorm2d(64)
        self.pool1 = nn.MaxPool2d(2,2)
        #self.dropout1 = nn.Dropout(0.25)
        self.dropout1 = nn.Dropout2d(0.025) # all 0.1

        # Conv block 2
        self.conv2 = nn.Conv2d(64, 128, kernel_size=(3,5), padding=(1,2))   # kernel_size=3, padding=1   wider now to see more time
        self.bn2 = nn.BatchNorm2d(128)
        self.pool2 = nn.MaxPool2d(2,2)
        #self.dropout2 = nn.Dropout(0.25)
        self.dropout2 = nn.Dropout2d(0.05)

        # Conv block 3
        self.conv3 = nn.Conv2d(128, 256, kernel_size=3, padding=1)
        self.bn3 = nn.BatchNorm2d(256)
        self.pool3 = nn.MaxPool2d(2,1)
        #self.dropout3 = nn.Dropout(0.25)
        self.dropout3 = nn.Dropout2d(0.05)

        # Conv block 4
        self.conv4 = nn.Conv2d(256, 512, kernel_size=3, padding=1)
        self.bn4 = nn.BatchNorm2d(512)
        self.pool4 = nn.MaxPool2d(1,2)
        #self.dropout4 = nn.Dropout(0.25)
        self.dropout4 = nn.Dropout2d(0.10)

        # global avg pooling
        self.GAP = nn.AdaptiveAvgPool2d((1,1))

        # fully connected layers
        self.fc1 = nn.Linear(512, 256)
        self.dropout_fc = nn.Dropout(0.5)
        self.fc2 = nn.Linear(256, num_classes)

    
    def forward(self, x):
        # Conv Block 1
        x = self.conv1(x)
        x = self.bn1(x)
        x = F.relu(x)
        x = self.pool1(x)
        x = self.dropout1(x)
        
        # Conv Block 2
        x = self.conv2(x)
        x = self.bn2(x)
        x = F.relu(x)
        x = self.pool2(x)
        x = self.dropout2(x)
        
        # Conv Block 3
        x = self.conv3(x)
        x = self.bn3(x)
        x = F.relu(x)
        x = self.pool3(x)
        x = self.dropout3(x)
        
        # Conv Block 4
        x = self.conv4(x)
        x = self.bn4(x)
        x = F.relu(x)
        x = self.pool4(x)
        x = self.dropout4(x)
        
        # Global Average Pooling
        x = self.GAP(x)
        x = torch.flatten(x, 1)     # x = x.view(x.size(0), -1)
        
        # Fully Connected
        x = self.fc1(x)
        x = F.relu(x)
        x = self.dropout_fc(x)
        x = self.fc2(x)
        
        return x


```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(device)

model = AudioCNN(num_classes=3, n_mels=64)
model = model.to(device)

# loss and optimizer
criterion = nn.CrossEntropyLoss()

optimizer = optim.Adam(model.parameters(), lr=0.0001) #0.001
#optimizer = optim.AdamW(model.parameters(), lr=0.0001, weight_decay=1e-4)
#optimizer = optim.SGD(model.parameters(), lr=0.01)

#scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=4, min_lr=1e-8)
scheduler = optim.lr_scheduler.OneCycleLR(optimizer, max_lr = 0.001, steps_per_epoch=len(train_loader), epochs=55)                                                               
```

```python
# training loop

AUGMENT = True

model_dir = "models"
os.makedirs(model_dir, exist_ok=True)

if AUGMENT:
    model_path = os.path.join(model_dir, "best_model_augmented_ONECYCLE_adam_drop2d2.pth")
else:
    model_path = os.path.join(model_dir, "best_model.pth")

epochs = 55
best_val_acc = 0.0

for epoch in range(epochs):

    model.train()
    total_train_loss = 0

    for x, y in train_loader:       # input, actual output

        x = x.to(device)
        y = y.to(device)

        # forward
        preds = model(x)
        loss = criterion(preds, y)

        # backward
        optimizer.zero_grad()
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()

        # scheduler step
        scheduler.step()

        total_train_loss += loss.item()


    model.eval()        # doesnt train
    total_val_loss = 0
    correct = 0
    total = 0

    with torch.no_grad():
        for x, y in val_loader:

            x = x.to(device)
            y = y.to(device)

            # forward
            preds = model(x)
            loss = criterion(preds, y)

            # no backward - just eval

            #   % acc
            _,predicted = torch.max(preds, 1)
            correct += (y == predicted).sum().item()
            total += y.size(0)

            total_val_loss += loss.item()

    val_accuracy = correct * 100 / total
    
    mean_train_loss = total_train_loss / len(train_loader)   # / num batches
    mean_val_loss = total_val_loss / len(val_loader)

    # scheduler step - 
    #scheduler.step(mean_val_loss)

    # Print current learning rate
    current_lr = optimizer.param_groups[0]['lr']
    print(f"Epoch {epoch+1}/{epochs}    Train Loss: {mean_train_loss:.4f}   Val Accuracy: {val_accuracy:.2f}%    Val Loss: {mean_val_loss:.4f}    LR: {current_lr:.7f}")


    # save best model
    if val_accuracy > best_val_acc:

        best_val_acc = val_accuracy
        torch.save(model.state_dict(), model_path)
        print(f"Saved new best model at epoch {epoch+1}\n")

```

```python
# for non augmented

import random
import numpy as np
from IPython.display import Audio, display

model = model.to(device)
model.eval()
class_names = list(test_dataset.classes)

def predict_from_index(idx: int):
    """Run the model on a single test_dataset item."""
    mel_db, label_idx = test_dataset[idx]          # (1, n_mels, T), label index
    x = mel_db.unsqueeze(0).to(device)             # (1, 1, n_mels, T)
    with torch.no_grad():
        logits = model(x)
        probs = torch.softmax(logits, dim=1).squeeze(0).cpu().numpy()
    pred_idx = int(probs.argmax())
    return int(label_idx), pred_idx, probs

def listen_from_test_loader(n=8, seed=None):
    """
    Pick n random samples from the test set (same data as test_loader),
    play the original audio, and show model predictions.
    """
    if seed is not None:
        random.seed(seed)

    indices = random.sample(range(len(test_dataset)), n)
    print(f"Random indices from test_dataset: {indices}")

    for idx in indices:
        row = test_dataset.df.iloc[idx]
        file_path = row["file_path"]
        true_label_str = row["class"]

        true_idx, pred_idx, probs = predict_from_index(idx)
        pred_label_str = class_names[pred_idx]

        print("\n" + "-"*60)
        print(f"Index: {idx}  File: {file_path}")
        print(f"True: {true_label_str}  |  Predicted: {pred_label_str}  (confidence {probs[pred_idx]:.2%})")

        # Play original audio
        display(Audio(file_path, rate=None))

        # Per-class probabilities
        order = np.argsort(probs)[::-1]
        for k in order:
            print(f"  {class_names[k]}: {probs[k]:.2%}")
```
